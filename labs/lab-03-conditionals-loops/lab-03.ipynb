### Conditional Statements:

Imagine you're a sports commentator. If a batsman scores 100 runs or more, you shout that they've made a century. Else if they score 50 or more, you announce a solid half-century. Else, you politely state they did neither.

That’s exactly how conditional statements work in Python! They let your code make decisions on the fly based on what's happening at that exact moment.

The main keywords for this are:

`if`: The main bouncer at the club. "If this condition is true, come on in!"

`elif`: The backup bouncer. "Okay, you didn't meet the first rule, but if you meet this rule, you can enter."

`else`: The safety net. "If all else fails, just do this."

Let's look at the cricket example in code:

```python
runs = int(input("How many runs did the batsman score? "))

if runs >= 100:
    print("Batsman scored a century! ")
elif runs >= 50:
    print("Batsman scored a fifty! ")
else:
    print("Batsman has neither scored a century nor fifty. Better luck next time!")
```

Fun Fact 🤓: You can stack as many `elif` blocks as you want! If you're calculating a store discount, you might check if sales >= 30000, then elif sales >= 20000, then elif sales >= 10000, all the way down to a default `else`!

The logic used in conditional statements (`True`/`False`) is called Boolean logic.

  ## Boolean Operators: The Logic Connectors 🔗

Boolean operators (`and`, `or`, `not`) are the glue that lets you combine multiple conditions. They only ever deal in absolutes: `True` or `False`.

### The `and` Operator
Both conditions must be `True`.  
**Analogy:** To launch a rocket, you need `fuel_is_full` AND `weather_is_clear`. If you are missing fuel or there's a storm, the launch is a no-go[cite: 1]. Both must be true!

### The `or` Operator
Only **one** condition needs to be `True`.  
**Analogy:** To get a museum discount, you can be an `is_student` OR an `is_senior_citizen`. You don't need to be both; just one gets you the deal!

### The `not` Operator
This is the "reverse" button—it simply flips the value. It turns `True` into `False`, and `False` into `True`.  
**Analogy:** If your condition is `not raining`, that just means it's a clear day. `not tired` means you are awake!

Pro-Tip: The Order of Logic 🤓
If you combine these, Python reads them in a specific order: $not$ is evaluated first, then $and$, and finally $or$. You can always use parentheses () to force a different order, just like in math!

## The `while` Loop & The Great Escape (`break`) 🏃‍♂️💨
A `while` loop is a conditional loop. It keeps repeating a block of code as long as a certain condition is `True`. But what if you need to bail out early because something went wrong?

Enter the `break` statement. It’s the emergency exit. If a `break` is triggered, Python drops whatever it's doing and immediately leaves the loop.

Imagine a program that asks for two numbers and divides them. If the user tries to divide by zero (which breaks the universe), we need to abort!



```python
# answer here
n1=int(input("Please plug in the dividend:"))
n2=int(input("Please plug in the divisor:"))

while n2 != 0:
  print(n1/n2)
  break
```

## The `for` Loop & The Art of Skipping (`continue`) 🦘

While `while` loops run until a condition changes, `for` loops are designed to run a specific number of times. They are perfect for counting!

But what if you are counting, and you just want to skip one specific iteration without destroying the whole loop? That's where `continue` comes in.

If `break` flips the dinner table and leaves the room, `continue` just skips the broccoli on its plate and moves on to the next bite.

Let's look at a loop that prints numbers from 1 to 10, but handles multiples of 3 very differently depending on the keyword used:



```python
#scenario A
for i in range(1, 11):
    if i % 3 == 0:
        break  # The loop DIES the moment i becomes 3.
    else:
        print(i)
# Output: 1, 2. (That's it. We went home.)
```

```python
#scenario B
for i in range(1, 11):
    if i % 3 == 0:
        continue  # Skips 3, 6, and 9, but keeps going!
    else:
        print(i)
# Output: 1, 2, 4, 5, 7, 8, 10
```

Pro-Tip: While `continue` is a cool jump trick, try not to overuse it! Sometimes simply writing a clean `if/else` block is much easier for humans to read.

```python
# Checkpoint 2: The Dramatic Ghost's Diary
quote = input("Enter a spooky ghost quote: ")
vowel_count = 0
vowels = "aeiouAEIOU"

for char in quote:
    if char in vowels:
        vowel_count += 1

if vowel_count > 10:
    print("Whoa, dial down the drama, Casper.")
else:
    print(f"Final vowel count: {vowel_count}")
```
